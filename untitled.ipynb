Build service given the necessary scopes

```python
from google.auth.transport.requests import Request
from google.oauth2.credentials import Credentials
from google_auth_oauthlib.flow import InstalledAppFlow
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError

import os.path

SCOPES = ['https://www.googleapis.com/auth/gmail.modify']

def get_token():
    creds = None
    token = None
    if os.path.exists("gcp/token.json"):
        creds = Credentials.from_authorized_user_file("gcp/token.json", SCOPES)
        print("Token loaded from gcp/token.json")
        
    if not creds or not creds.valid:
        if creds and creds.expired and creds.refresh_token:
            creds.refresh(Request())
        else:
            flow = InstalledAppFlow.from_client_secrets_file(
                "gcp/credentials.json", SCOPES
            )
            creds = flow.run_local_server(port=0)
            # Save the credentials for the next run
        with open("gcp/token.json", "w") as token:
            token.write(creds.to_json())
            print("Token saved to gcp/token.json")
    return token

get_token()
```

```python
service = build("gmail", "v1", credentials=Credentials.from_authorized_user_file("gcp/token.json", SCOPES))
```

```python
# this one fetches the latest email

def get_latest_email(service):
    try:
        response = service.users().messages().list(userId='me', maxResults=1).execute()
        messages = response.get('messages', [])
        if not messages:
            print("no messages found")
            return None

        latest_message_id = messages[0]['id']    
        full_message = service.users().messages().get(
            userId='me', 
            id=latest_message_id
        ).execute()
        
        msg_history_id = full_message.get('historyId')
        headers = full_message['payload']['headers']
        subject = next(h['value'] for h in headers if h['name'] == 'Subject')
        
        print(f"Message ID: {latest_message_id}")
        print(f"Subject: {subject}")
        print(f"Message History ID: {msg_history_id}")
        return full_message
        
    except Exception as error:
        print(f"An error occurred: {error}")
        return None
    
latest_email = get_latest_email(service)
```

```python
# this one fetches the first message from the list given history_id
import base64

def get_message(service, history_id):
    try:
        history_response = service.users().history().list(
            userId='me', startHistoryId=history_id, historyTypes=['messageAdded']
            ).execute()
        changes = history_response.get('history', [])
        print(f"changes: {changes}")
        if not changes:
            print("No new emails")
            return

        for record in changes:
            added_messages = record.get('messagesAdded', [])
            
            for msg_item in added_messages:
                message_id = msg_item['message']['id']
                
                msg = service.users().messages().get(userId='me', id=message_id).execute()

                print(f"Full message data: {msg}")
                
                headers = msg['payload']['headers']
                subject = next(h['value'] for h in headers if h['name'] == 'Subject')
                sender = next(h['value'] for h in headers if h['name'] == 'From')
                print(f"Email found: {subject} from {sender}")
                print(f"Message ID: {message_id}")
                parts = msg['payload'].get('parts', [])
                print(len(parts))

                for part in parts:
                    if part['mimeType'] == 'text/plain':
                        return part
                # print(f"Message payload: {msg['payload']['parts']}")

                return None
                
    except Exception as error:
        print(f'An error occurred: {error}')

temp = get_message(service, latest_email['historyId'])
print(temp)

body_data = temp['body']['data']
body_text = base64.urlsafe_b64decode(body_data).decode('utf-8')
# print(body_text)
```

```python
# body_text

import re
from bs4 import BeautifulSoup

def clean_email_body(raw_content):
    """
    Removes HTML, CSS, hidden characters, shortens URLs.
    """
    if not raw_content:
        return ""

    # html tags and css/style blocks
    soup = BeautifulSoup(raw_content, "html.parser")
    
    # <style>, <script>, and <link> tags
    for element in soup(["style", "script", "link"]):
        element.decompose()

    text = soup.get_text(separator=' ')
    text = text.replace('\u200c', '').replace('\xa0', ' ')

    # url handling
    def url_cleaner(match):
        url = match.group(0)
        if len(url) > 60:
            return "[URL]"
        return url

    # re.sub(pattern, repl, string)
    # r'https?://\S+' - either http or https followed by characters until space.
    # matched text is passed to url_cleaner
    text = re.sub(r'https?://\S+', url_cleaner, text)

    # replaces multiple newlines, spaces with a single one
    text = re.sub(r'\n\s*\n', '\n', text)
    text = re.sub(r' +', ' ', text)
    text = text.strip()

    return text

cleaned_text = clean_email_body(body_text)
print(cleaned_text)
```

Also, we should (manually) check the size estimate - if the message is super large, we want to avoid send its content to an LLM provider too so we do not incur huge costs.

Fields without *too much* sensitive data that we can use to filter by:
- subject
- from
- size estimate

What to avoid *initially*:
- snippet (may contain temporary password reset link / reset code)
- encoded text content

If an LLM says that a given email is "safe" then we can ask it to categorize it and give it a label. If not, we want to add it a different label - like "might be sensitive". If the estimated size is super huge, we want to label it as "too large". We need to label those messages so that we avoid looping over them indifinitely.

```python
def get_label_map(service):
    """dictionary of {label_name: label_id}."""
    results = service.users().labels().list(userId='me').execute()
    labels = results.get('labels', [])
    
    label_map = {label['name']: label['id'] for label in labels}
    return label_map

label_map = get_label_map(service)
print(label_map)
```

```python
# Label_4550656612000604041

def get_new_message_ids(service, start_history_id):
    """list of message IDs added since the history_id."""
    history = service.users().history().list(
        userId='me', 
        startHistoryId=start_history_id,
        historyTypes=['messageAdded']
    ).execute()
    message_ids = []
    if 'history' in history:
        for h in history['history']:
            for msg_added in h.get('messagesAdded', []):
                message_ids.append(msg_added['message']['id'])
    return list(set(message_ids)) # avoid duplicates

temp_msg = get_new_message_ids(service, latest_email['historyId'])
```

```python
def fetch_if_not_processed(service, message_id, ai_label_id):
    """Fetches message if it doesn't have the AI_PROCESSED label."""
    msg = service.users().messages().get(userId='me', id=message_id).execute()
    
    # Check current labels
    current_labels = msg.get('labelIds', [])
    print(current_labels)
    if ai_label_id in current_labels:
        return None # Already processed
    
    # Extract snippet for LLM
    subject = next(h['value'] for h in msg['payload']['headers'] if h['name'] == 'Subject')
    snippet = msg.get('snippet', '')
    
    return {'id': message_id, 'text': f"Subject: {subject}\nContent: {snippet}"}

fetch_if_not_processed(service, '19c72344dc2d14f1', 'Label_4550656612000604041')
```

```python
def process_and_categorize(service, message_id, ai_label_id, category_label_id):

    body = {
        'addLabelIds': [ai_label_id, category_label_id],
        'removeLabelIds': ['INBOX'] # Archive it once sorted
    }
    
    service.users().messages().modify(
        userId='me', 
        id=message_id, 
        body=body
    ).execute()

process_and_categorize(service, '19c72344dc2d14f1', 'Label_4550656612000604041', 'Label_6075941912103430332')
```
